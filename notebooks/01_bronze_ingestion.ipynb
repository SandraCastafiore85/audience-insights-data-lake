```python
import pandas as pd


# ingest data
ratings = pd.read_csv("../data/raw/ratings.csv")
movies = pd.read_csv("../data/raw/movies.csv")
tags = pd.read_csv("../data/raw/tags.csv")
links = pd.read_csv("../data/raw/links.csv")

# add metadata
ratings["ingestion_ts"] = pd.Timestamp.now(tz="UTC")
ratings["source_file"] = "ratings.csv"
movies["ingestion_ts"] = pd.Timestamp.now(tz="UTC")
movies["source_file"] = "movies.csv"
tags["ingestion_ts"] = pd.Timestamp.now()
tags["source_file"] = "tags.csv"
links["ingestion_ts"] = pd.Timestamp.now(tz="UTC")
links["source_file"] = "links.csv"
```

```python
#inspect
pd.set_option('display.width', 300)
print("Ratings:")
print(ratings.head())
print("Movies:")
print(movies.head())
print("Tags:")
print(tags.head())
print("Links:")
print(links.head())
```

```python
# basic data quality checks
# row counts
print(f"ratings length: {len(ratings)}")
print(f"movies length: {len(movies)}")
print(f"tags length: {len(tags)}")
print(f"links length: {len(links)}")


# check for duplicates
try:
    assert ratings.duplicated().sum() == 0, "Expected no duplicates in ratings"
    print("Success: No duplicate rows found in ratings.")
except AssertionError as e:
    print(f"Failure: {e}")
try:
    assert movies.duplicated().sum() == 0, "Expected no duplicates in movies"
    print("Success: No duplicate rows found in movies.")
except AssertionError as e:
    print(f"Failure: {e}")
try:
    assert tags.duplicated().sum() == 0, "Expected no duplicates in tags"
    print("Success: No duplicate rows found in tags.")
except AssertionError as e:
    print(f"Failure: {e}")
try:
    assert links.duplicated().sum() == 0, "Expected no duplicates in links"
    print("Success: No duplicate rows found in links.")
except AssertionError as e:
    print(f"Failure: {e}")
```

```python
# save as Parquet files
ratings.to_parquet("../data_processed/bronze/ratings.parquet", index=False)
movies.to_parquet("../data_processed/bronze/movies.parquet", index=False)
tags.to_parquet("../data_processed/bronze/tags.parquet", index=False)
```
